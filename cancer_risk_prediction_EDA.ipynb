```python
## Finding out which factors highly contribute to high risk level

# Select only numerical columns
numeric_columns = df.select_dtypes(include='number').columns

# Compute the mean of each numeric feature per risk level
risk_means = df.groupby('Risk_Level')[numeric_columns].mean().T

print(risk_means)
```

# Insights

- Heavy smokers are much more likely to be High risk.
- Similar to smoking – high alcohol use strongly linked to High risk.
- High red meat consumption is associated with High risk.
- High intake of salted/processed foods strongly linked to High risk.
- Exposure to high air pollution is a major differentiator for High risk.
- Jobs with hazardous exposures push risk higher.
- Higher obesity levels correlate with higher risk.

```python
# Focus on difference between High and Low Risk
risk_diff = risk_means['High'] - risk_means['Low']
risk_diff = risk_diff.sort_values(ascending=False)

print(risk_diff)
```

```python
risk_diff.sort_values(ascending=True).plot(kind='barh', figsize=(8,6), title='Top Factors Contributing to High Risk Level')
plt.xlabel('Difference (High - Low)')
plt.ylabel('Feature')
plt.show()
```

```python
bmi_by_risk = df.groupby('Risk_Level')['BMI'].mean()
print(bmi_by_risk)
```

## **Univariate Analysis**

```python
for i, predictor in enumerate(df.drop(columns=['Patient_ID', 'Risk_Level', 'Age', 'Overall_Risk_Score', 'BMI'])):
    plt.figure(i)
    sns.countplot(data=df, x=predictor, hue='Risk_Level')
```

```python


# Define consistent color palette and order
risk_palette = {
    'High': 'green',
    'Medium': 'blue',
    'Low': 'orange'
}
risk_order = ['Low', 'Medium', 'High']  # Controls legend order (bottom to top)

# Your loop with fixed colors
predictors = df.drop(columns=['Patient_ID', 'Risk_Level', 'Age', 'Overall_Risk_Score', 'BMI'])

for i, predictor in enumerate(predictors):
    plt.figure(i, figsize=(10, 6))
    sns.countplot(
        data=df,
        x=predictor,
        hue='Risk_Level',
        palette=risk_palette,
        order=sorted(df[predictor].unique()),  # optional: sort x-axis
        hue_order=risk_order
    )
    plt.title(f'{predictor} vs Risk Level')
    plt.xticks(rotation=45)
    plt.legend(title='Risk Level')
    plt.tight_layout()
    plt.show()
```

## Numerical Analysis

```python
# Consistent color palette (matching your previous countplots)
risk_palette = {
    'High': 'green',
    'Medium': 'blue',  # Gold/yellow for better visibility
    'Low': 'orange'
}

# Optional: control legend order
risk_order = ['Low', 'Medium', 'High']

plt.figure(figsize=(10, 6))
sns.kdeplot(
    data=df,
    x='Age',
    hue='Risk_Level',
    palette=risk_palette,
    hue_order=risk_order,
    fill=True,          # Shades under the curves for better visual distinction
    alpha=0.6,          # Slight transparency so overlaps are visible
    linewidth=2
)

plt.title('Distribution of Age by Risk Level', fontsize=14)
plt.xlabel('Age')
plt.ylabel('Density')
plt.legend(title='Risk Level')
plt.tight_layout()
plt.show()
```
